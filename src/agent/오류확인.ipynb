```python
from tools.text2cypher import search_graph

result = search_graph("""
MATCH (a)-[r]->(b)
RETURN
    a.name AS source,
    labels(a)[0] AS source_type,
    type(r) AS relationship,
    b.name AS target,
    labels(b)[0] AS target_type,
    r.evidence AS evidence,
    r.source_case AS source_case,
    r.source_row AS source_row
LIMIT 10
""")

print(result)
```
