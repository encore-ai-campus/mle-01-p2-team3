```python
import pickle
from pathlib import Path
from IPython.display import JSON, display

# 경로 설정
PKL_PATH = Path("../data/light_data.pkl")

with open(PKL_PATH, "rb") as f:
    data = pickle.load(f)

display(JSON(data[0], expanded=True))
```

```python
from collections import Counter
import pandas as pd

# 설정
TOP_N_AGENTS = 10

# (name, persona) 조합으로 집계
agent_counts = Counter(
    (agent["name"], agent["persona"])
    for episode in data
    for agent in episode["agents"]
)

# 상위 N명
top_agents = pd.DataFrame(
    [
        {
            "name": name,
            "persona": persona,
            "count": count
        }
        for (name, persona), count
        in agent_counts.most_common(TOP_N_AGENTS)
    ]
)

display(top_agents)
```

```python
# 상위 agent의 (name, persona) 집합
selected_agents = set(
    zip(top_agents["name"], top_agents["persona"])
)

episode_rows = []

for idx, episode in enumerate(data):

    # 해당 에피소드에 등장하는 (name, persona)
    agents = {
        (agent["name"], agent["persona"])
        for agent in episode["agents"]
    }

    # 모든 등장 agent가 상위 N명에 포함되어야 함
    if agents and agents.issubset(selected_agents):
        episode_rows.append({
            "episode": idx,
            "agents": [agent["name"] for agent in episode["agents"]],
            "setting": episode["setting"]["name"]
        })

selected_episodes = pd.DataFrame(episode_rows)

display(selected_episodes)
```
