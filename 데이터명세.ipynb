| 항목 | 내용 |
| --- | --- |
| **파일명** | `data/clean/모기업_계열사_종속기업_통합.csv` |
| **행 × 열** | **17,398행 × 17열** |
| **기간** | 기준일·기준연도 컬럼이 없어 **알 수 없음** |

| **결측·중복** | 완전 중복 0건, 식별 키 조합 중복 0건<br>관계 유형에 따라 사용하지 않는 컬럼은 구조적으로 비어 있음<br>주소·지역·업종 컬럼에는 원천 데이터 부족으로 인한 실제 결측도 존재 |

## 3. 컬럼 명세서

> 척도는 명목·서열·등간·비율척도 중에서 구분합니다. 이 데이터셋의 컬럼은 관계 유형, 식별자, 이름, 주소 등 범주 정보이므로 모두 **명목척도**입니다.

| 컬럼명 | 설명 | 자료형 | 척도 | 값·형식 | 결측 | 고유값 | 예시·주의사항 |
| --- | --- | --- | --- | --- | ---: | ---: | --- |
| `case` | 한 행이 나타내는 기업 관계 유형 | 범주형 정수 | 명목 | `1`: 모기업→계열사<br>`2`: 모기업→종속기업<br>`3`: 모기업→계열사→종속기업 | 0건<br>0.00% | 3 | 예시: `1`<br>값의 크기나 순서에는 의미가 없음 |
| `top_crno` | 최상위 모기업 법인등록번호 | 문자열 | 명목 | 개별 번호는 13자리 숫자 문자열<br>`case=3`에서는 복수 번호를 `;`로 연결 | 0건<br>0.00% | 원문 2,118<br>개별 번호 1,724 | 예시: `1101110000086`<br>숫자형으로 변환하지 않으며 다른 `top_*` 컬럼과 같은 위치 기준으로 분리 |
| `top_corpNm` | 최상위 모기업명 | 문자열 | 명목 | 단일 기업명 또는 `;`로 연결된 복수 기업명 | 0건<br>0.00% | 2,112 | 예시: `롯데쇼핑(주)`<br>복수값은 `top_crno`와 동일한 순서로 대응 |

| `top_sicNm` | 모기업 업종명 | 문자열 | 명목 | 기업개요 원천의 `sicNm`을 사용하고 없으면 `enpMainBizNm` 사용 | 11,134건<br>64.00% | 306 | 예시: `백화점`<br>case 1 결측률 82.18%, case 2 52.48%, case 3 38.90% |
