## 0. 불러오기

```python
import pandas as pd
from pathlib import Path

df = pd.read_csv(Path("../data/merged/종속기업_원본.csv"), dtype=str)
print(f"{len(df):,}행")
```

## 1. 기본 점검

```python
print("완전 중복 행:", df.duplicated().sum())
print("crno 13자리 아님:", (df["crno"].str.len() != 13).sum())
print("모회사 수:", df["crno"].nunique())
print("이름 '-' 또는 빈칸:", df["sbrdEnpNm"].fillna("").str.strip().isin(["", "-"]).sum())
```

## 2. 기준일 점검

```python
multi = (df.groupby("crno")["basDt"].nunique() > 1).sum()
print("기준일 2개 이상인 모회사:", multi)
print(df["basDt"].value_counts().head())
```
