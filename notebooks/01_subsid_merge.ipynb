## 0. 파일 병합 (종속기업)

```python
import unicodedata
import pandas as pd
from pathlib import Path

RAW_DIR = Path("../금융위원회_기업기본정보_수집")
MERGED_DIR = Path("../data/merged")
MERGED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# 맥에서 한글 파일명이 안 찾아지는 문제 방지
files = sorted(
    f for f in RAW_DIR.glob("*.csv")
    if unicodedata.normalize("NFC", f.name).startswith("연결대상종속기업")
)
print([f.name for f in files])

subsid = pd.concat(
    [pd.read_csv(f, dtype=str) for f in files],
    ignore_index=True,
)
print(f"파일 {len(files)}개, {len(subsid):,}행")
```

```python
subsid.to_csv(MERGED_DIR / "종속기업_원본.csv", index=False, encoding="utf-8-sig")
```
